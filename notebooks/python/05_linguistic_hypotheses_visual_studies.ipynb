### Executive Overview
This notebook conducts visual and quantitative evaluations of three core hypotheses in cultural and linguistic evolution:
1. **Hypothesis 6 (Representational Economics)**: *Winner-Take-All Notational Extinction in Mathematics vs. Polyglot Coexistence in Software Syntax*. We examine the historical displacement of Newton's fluxional dot-notation by Leibnizian differentials post-1812 alongside the multi-paradigm coexistence of programming language syntax (C braces, Python indentation, Lisp S-expressions).

We model diachronic decay rates, contrasting the rapid pejoration and 15–25 year obsolescence of top-down institutional euphemisms against the multi-decade resilience (40–90+ years) of subcultural enantiosemic inversions (*cool*, *ill*, *dope*, *bad*).

## 2. Hypothesis 6: Representational Economics (Extinction vs. Polyglot Coexistence)
- **Mathematics (1700–1900)**: Following the Newton–Leibniz priority dispute (1712), British mathematics institutionalized Newtonian fluxions ($\dot{x}, \ddot{x}$), while continental Europe adopted Leibnizian differentials ($\frac{dx}{dt}, \int$). By 1812, the Cambridge Analytical Society triggered an immediate collapse of fluxional notation in favor of the operational Leibnizian standard.
- **Software Systems (1975–2025)**: Programming language syntax avoided global monopolization. C-style curly braces (`{}`), off-side rule indentation (Python), and S-expressions (Lisp) coexist because they optimize orthogonal operational constraints.


```python
# Load empirical time series
df_not = pd.read_csv("../../data/corpus/notational_extinction.csv")
df_not.head(8)
```

```python
# Visualizing Representational Economics in Mathematics vs. Software
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(11, 10), facecolor=bg_color)
for ax in (ax1, ax2):
    ax.set_facecolor(bg_color)
    ax.grid(True, color='#E2E8F0', linestyle='-', linewidth=0.6, zorder=0)

# Panel A: Mathematics
df_math = df_not[df_not['system'] == 'Mathematics']
for notation, color in [("Newtonian Fluxions (x-dot)", "#C2410C"), ("Leibnizian Differentials (dy/dx)", "#1E3A8A")]:
    sub = df_math[df_math['notation'] == notation]
    ax1.plot(sub['epoch'], sub['share_pct'], marker='o', linewidth=2.4, markersize=7, 
             label=notation, color=color, zorder=3)

ax1.set_title("A. Mathematical Calculus (1700–1900): Winner-Take-All Notational Extinction", 
              fontsize=13, fontweight='bold', color=text_color, pad=10, loc='left')
ax1.set_ylabel("Literature & Pedagogy Share (%)", fontsize=10, fontweight='bold', color=text_color)
ax1.set_ylim(0, 105)
ax1.yaxis.set_major_formatter(ticker.PercentFormatter())
ax1.legend(frameon=True, facecolor=bg_color, edgecolor='#CBD5E1', fontsize=9)

ax1.annotate("1812: Cambridge Analytical Society\n'Pure d-ism vs. the dot-age'", 
             xy=(1812, 14), xytext=(1760, 35),
             arrowprops=dict(arrowstyle="->", color="#9A3412", lw=1.2),
             bbox=dict(boxstyle="round,pad=0.4", facecolor="#FFF7ED", edgecolor="#9A3412", lw=0.8),
             fontsize=9, fontweight='bold', color="#9A3412")

# Panel B: Software
df_soft = df_not[df_not['system'] == 'Programming Syntax']
colors = {
    "C-Style Braces ({})": "#0284C7",
    "Off-side Whitespace (Python/Haskell)": "#059669",
    "S-Expressions (Lisp)": "#7C3AED",
    "Algol/Pascal Keywords (begin/end)": "#D97706"
}
for notation, col in colors.items():
    sub = df_soft[df_soft['notation'] == notation]
    ax2.plot(sub['epoch'], sub['share_pct'], marker='o', linewidth=2.2, markersize=6, 
             label=notation, color=col, zorder=3)

ax2.set_title("B. Software Syntax (1975–2025): Polyglot Multi-Paradigm Coexistence", 
              fontsize=13, fontweight='bold', color=text_color, pad=10, loc='left')
ax2.set_xlabel("Year (Epoch)", fontsize=10, fontweight='bold', color=text_color)
ax2.set_ylabel("Active Repositories Share (%)", fontsize=10, fontweight='bold', color=text_color)
ax2.set_ylim(0, 80)
ax2.yaxis.set_major_formatter(ticker.PercentFormatter())
ax2.legend(frameon=True, facecolor=bg_color, edgecolor='#CBD5E1', fontsize=8.5, loc='upper left')

plt.suptitle("REPRESENTATIONAL ECONOMICS: EXTINCTION VS. POLYGLOT COEXISTENCE", 
             fontsize=15, fontweight='bold', color='#0F172A', y=0.98)
plt.tight_layout(rect=[0, 0.03, 1, 0.96])
plt.show()
```

```python
# Quantitative Summary of Representational Economics
for sys, grp in df_not.groupby('system'):
    latest_epoch = grp['epoch'].max()
    latest_grp = grp[grp['epoch'] == latest_epoch]
    shares = latest_grp['share_pct'].values / 100.0
    entropy = -np.sum(shares * np.log(shares))
    print(f"System: {sys} | Final Year: {latest_epoch} | Leading Syntax: {latest_grp.loc[latest_grp['share_pct'].idxmax(), 'notation']} ({latest_grp['share_pct'].max()}%) | Shannon Entropy: {entropy:.3f}")
```

## 3. Hypothesis 5: Lexical Density & The Working Vocabulary Horizon
- **Methodology**: Following Matt Daniels (The Pudding), we analyze unique word counts within the first **35,000 words** of an artist's discography or author's prose corpus.
- **Key Benchmarks**:
  - Herman Melville (*Moby Dick*): **6,022 unique words**
  - William Shakespeare (Sonnets & Plays): **5,170 unique words**
  - Genre Median (Hip-Hop): **~4,200 unique words**


```python
# Load Vocabulary Horizon Dataset
df_vocab = pd.read_csv("../../data/corpus/vocabulary_horizon.csv")
df_vocab = df_vocab.sort_values('unique_words', ascending=True).reset_index(drop=True)
df_vocab.tail(8)
```

```python
# Visualize Working Vocabulary Spectrum
fig, ax = plt.subplots(figsize=(12, 10), facecolor=bg_color)
ax.set_facecolor(bg_color)
ax.grid(axis='x', color='#E2E8F0', linestyle='--', linewidth=0.6, zorder=0)

y_pos = np.arange(len(df_vocab))
colors_cat = []
for cat in df_vocab['category']:
    if 'Classical' in cat:
        colors_cat.append('#DC2626')
    elif 'Underground' in cat:
        colors_cat.append('#7C3AED')
    elif 'Golden Era' in cat:
        colors_cat.append('#D97706')
    elif 'Southern' in cat:
        colors_cat.append('#059669')
    else:
        colors_cat.append('#2563EB')

ax.hlines(y=y_pos, xmin=3000, xmax=df_vocab['unique_words'], color=colors_cat, alpha=0.4, linewidth=1.5)
scatter = ax.scatter(df_vocab['unique_words'], y_pos, c=colors_cat, s=60, zorder=3)

for i, row in df_vocab.iterrows():
    ax.text(row['unique_words'] + 60, i, f"{row['unique_words']:,}", va='center', 
            fontsize=8.5, fontweight='bold', color=text_color)

ax.axvline(5170, color='#991B1B', linestyle='--', linewidth=1.2, zorder=2)
ax.text(5170, 2, " Shakespeare Benchmark (5,170 words)", color='#991B1B', fontweight='bold', fontsize=9)

ax.axvline(6022, color='#92400E', linestyle=':', linewidth=1.2, zorder=2)
ax.text(6022, 4, " Herman Melville: Moby Dick (6,022 words)", color='#92400E', fontweight='bold', fontsize=9)

ax.set_yticks(y_pos)
ax.set_yticklabels(df_vocab['artist_author'], fontsize=9, fontweight='bold')
ax.set_xlim(2900, 8100)
ax.set_xlabel("Unique Words in First 35,000 Words of Lyrics / Prose", fontsize=10, fontweight='bold', color=text_color)
ax.set_title("THE WORKING VOCABULARY HORIZON: HIP-HOP LYRICISM VS. LITERARY CANON", 
             fontsize=14, fontweight='bold', color='#0F172A', pad=15, loc='left')

plt.tight_layout()
plt.show()
```

## 4. Hypothesis 4: Semantic Durability (Euphemism Treadmill vs. Vernacular Shields)
- **The Euphemism Treadmill (Pinker, 1994, 2002)**: Bureaucratic substitutions decay within 15–25 years as underlying stigma attaches to the new signifier (*imbecile* $\rightarrow$ *mentally retarded* $\rightarrow$ *special needs* $\rightarrow$ *intellectually disabled*).
- **Organic Vernacular Shields**: Subcultural enantiosemic inversions (*cool*, *bad*, *ill*, *dope*) demonstrate extraordinary diachronic half-lives (40–90+ years) while maintaining high prestige.


```python
# Load Euphemism and Vernacular Durability Data
df_tread = pd.read_csv("../../data/corpus/euphemism_treadmill.csv")
df_tread.head(8)
```

```python
# Phase Space Visualization: Half-Life Stability vs. Contemporary Social Valence
fig, ax = plt.subplots(figsize=(11, 7.5), facecolor=bg_color)
ax.set_facecolor(bg_color)
ax.grid(True, color='#E2E8F0', linestyle='-', linewidth=0.6, zorder=0)

ax.axhline(0, color='#94A3B8', linestyle='--', linewidth=0.8)
ax.axvline(30, color='#CBD5E1', linestyle=':', linewidth=0.8)

# Shaded zones
ax.axvspan(10, 30, ymin=0.02, ymax=0.65, color='#DC2626', alpha=0.06)
ax.text(20, -0.92, "TREADMILL TRAP\nRapid Pejoration & Replacement", 
        color='#991B1B', fontweight='bold', fontsize=9.5, ha='center')

ax.axvspan(35, 102, ymin=0.72, ymax=0.98, color='#059669', alpha=0.06)
ax.text(68, 1.05, "VERNACULAR SHIELD\nMulti-Decade Subcultural Prestige Armor", 
        color='#065F46', fontweight='bold', fontsize=9.5, ha='center')

for cat, col in [("Institutional Euphemism", "#DC2626"), ("Organic Vernacular Shield", "#059669")]:
    sub = df_tread[df_tread['category'] == cat]
    ax.scatter(sub['half_life_years'], sub['current_valence'], 
               color=col, s=(sub['current_valence'] + 1.2) * 50, alpha=0.85, label=cat, zorder=3)
    for _, r in sub.iterrows():
        nudge = -0.06 if cat == "Institutional Euphemism" else 0.05
        ax.text(r['half_life_years'], r['current_valence'] + nudge, f"{r['term']} ({r['year_introduced']})",
                fontsize=8, fontweight='bold', color=col, ha='center')

ax.set_xlim(10, 105)
ax.set_ylim(-1.05, 1.15)
ax.set_xlabel("Linguistic Half-Life (Years of Semantic Stability)", fontsize=10, fontweight='bold', color=text_color)
ax.set_ylabel("Contemporary Social Valence Score", fontsize=10, fontweight='bold', color=text_color)
ax.set_title("THE EUPHEMISM TREADMILL VS. ORGANIC VERNACULAR SHIELDS", 
             fontsize=14, fontweight='bold', color='#0F172A', pad=15, loc='left')
ax.legend(frameon=True, facecolor=bg_color, edgecolor='#CBD5E1', fontsize=9, loc='center left')

plt.tight_layout()
plt.show()
```

```python
# Comparative Statistics
df_tread.groupby('category').agg(
    count=('term', 'count'),
    mean_half_life=('half_life_years', 'mean'),
    median_half_life=('half_life_years', 'median'),
    min_half_life=('half_life_years', 'min'),
    max_half_life=('half_life_years', 'max'),
    mean_valence=('current_valence', 'mean')
).round(2)
```

### Q&A
- **Q: Why does mathematical notation experience total extinction while programming language syntax sustains polyglot coexistence?**  
  **A:** Mathematical notation is bound by a singular global communicational medium (proof publication and unified theorem exchange), rendering incompatible syntaxes zero-sum. Once the Cambridge Analytical Society adopted Leibnizian differentials in 1812, Newtonian fluxions experienced rapid obsolescence (plummeting from a 48% share in 1700 to near 0% for general calculus by 1840). In contrast, software languages decouple execution domains: C-family braces (56% in 2025) dominate system-level deterministic parsing, Python significant whitespace (31%) minimizes cognitive friction in data science and AI orchestration, and Lisp S-expressions (7%) provide homoiconic macro-metaprogramming.
- **Q: Does classical literature represent the upper bound of English lexical diversity?**  
  **A:** No. In a standardized sample of the first 35,000 words of artistic output, underground and golden era hip-hop lyricists systematically surpass classical benchmarks. Aesop Rock records 7,392 unique words (+2,222 unique words, or +43.0% over William Shakespeare's 5,170 words, and +1,370 words over Herman Melville's 6,022 words). Four distinct hip-hop artists (Aesop Rock, GZA, Kool Keith, MF DOOM) exceed Herman Melville's *Moby Dick*, demonstrating that multi-syllabic rhyming structures enforce intense lexical expansion.
- **Q: Why do top-down euphemisms fail to eradicate stigma while subcultural vernacular slang endures for generations?**  
  **A:** Steven Pinker's euphemism treadmill reveals that stigma is conceptual rather than phonetic: institutional substitutes decay within an observed mean half-life of 20.8 years before acquiring pejorative charge (*imbecile* $\rightarrow$ *mentally retarded* $\rightarrow$ *special needs* $\rightarrow$ *intellectually disabled*). Conversely, subcultural enantiosemic inversions (*cool*, *bad*, *ill*, *dope*) demonstrate an observed mean half-life of 50.8 years (reaching 93 years for *cool*) because syntactic and cultural barriers insulate the inverted prestige signifier from bureaucratic dilution.

### Data Analysis Key Findings
- **Shannon Entropy of Syntactic Diversity**: Mathematics collapsed from a high-entropy competitive state in 1700 ($H = 0.692$) to a near-zero entropy monopoly by 1900 ($H = 0.098$, with Leibnizian notation commanding 98.0% of calculus literature). In software syntax, entropy has remained sustainably high across 50 years ($H = 1.054$ in 2025), preserving a four-paradigm ecosystem.
- **Underground Lyricism Threshold**: 10 out of 25 sampled hip-hop discographies exceed William Shakespeare's canonical working vocabulary of 5,170 words per 35,000-word corpus. The underground abstract tradition averages 6,599 unique words, outperforming both the genre median (4,200 words) and canonical English literature.
- **Half-Life Disparity in Lexical Engineering**: Institutional euphemisms exhibit a maximum observed half-life of only 28.0 years (*underprivileged* and *neurodivergent*), with a mean contemporary social valence of $-0.291$. In contrast, organic vernacular shields exhibit a minimum half-life of 21.0 years (*fire*) and an average of 50.8 years, retaining a strongly positive mean valence of $+0.861$.
