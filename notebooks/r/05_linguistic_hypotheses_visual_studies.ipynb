## 2. Hypothesis 6: Representational Economics (Extinction vs. Polyglot Coexistence)

### Theoretical Context
In formal systems, does symbolic competition follow a winner-take-all monopolistic trajectory, or does it reach a sustainable polyglot equilibrium?
- **Mathematics (1700–1900)**: Following the Newton–Leibniz priority dispute (1712), British mathematics institutionalized Newtonian fluxional notation ($\dot{x}, \ddot{x}$), while continental Europe adopted Leibnizian differentials ($\frac{dx}{dt}, \int$). By 1812, the Cambridge Analytical Society (Charles Babbage, John Herschel, George Peacock) advocated for "pure d-ism in opposition to the dot-age of the university," causing an immediate collapse of fluxional notation in favor of the operational Leibnizian standard.
- **Software Systems (1975–2025)**: Conversely, programming language syntax has avoided global monopolization. C-style curly braces (`{}`), off-side rule indentation (Python), and S-expressions (Lisp) coexist because they optimize orthogonal operational frontiers (parser predictability, visual scanning speed, and macro homoiconicity).


```r
# Load empirical time series
df_not <- read.csv("../../data/corpus/notational_extinction.csv")
head(df_not, 8)
```

```r
# Construct Comparative Multi-Panel Study
p_math <- df_not %>%
  filter(system == "Mathematics") %>%
  ggplot(aes(x = epoch, y = share_pct, color = notation, group = notation)) +
  geom_line(linewidth = 1.3) +
  geom_point(size = 3.2, fill = "#FAF8F5", shape = 21, stroke = 1.5) +
  scale_color_manual(values = c("Newtonian Fluxions (x-dot)" = "#C2410C", 
                                "Leibnizian Differentials (dy/dx)" = "#1E3A8A")) +
  scale_x_continuous(breaks = c(1700, 1740, 1780, 1812, 1840, 1900)) +
  scale_y_continuous(labels = function(x) paste0(x, "%"), limits = c(0, 108)) +
  labs(
    title = "A. Mathematical Calculus (1700–1900): Winner-Take-All Notational Extinction",
    subtitle = "Newton's dot-notation held monopoly in the UK until the 1812 Cambridge Analytical Society triggered total collapse.",
    x = "Year (Epoch)",
    y = "Literature & Pedagogy Share",
    color = "Notational System"
  ) +
  theme_editorial() +
  annotate("label", x = 1812, y = 20, label = "1812: Cambridge Analytical Society\n'Pure d-ism vs. the dot-age'", 
           size = 2.8, fill = "#FFF7ED", color = "#9A3412", fontface = "bold", hjust = 0.5) +
  annotate("label", x = 1880, y = 96, label = "Global Standard:\nLeibnizian Differentials", 
           size = 2.8, fill = "#EFF6FF", color = "#1E3A8A", fontface = "bold", hjust = 0.5)

p_soft <- df_not %>%
  filter(system == "Programming Syntax") %>%
  ggplot(aes(x = epoch, y = share_pct, color = notation, group = notation)) +
  geom_line(linewidth = 1.2) +
  geom_point(size = 3.0, fill = "#FAF8F5", shape = 21, stroke = 1.4) +
  scale_color_manual(values = c(
    "C-Style Braces ({})" = "#0284C7",
    "Off-side Whitespace (Python/Haskell)" = "#059669",
    "S-Expressions (Lisp)" = "#7C3AED",
    "Algol/Pascal Keywords (begin/end)" = "#D97706"
  )) +
  scale_x_continuous(breaks = c(1975, 1990, 2005, 2025)) +
  scale_y_continuous(labels = function(x) paste0(x, "%"), limits = c(0, 80)) +
  labs(
    title = "B. Software Syntax (1975–2025): Polyglot Multi-Paradigm Coexistence",
    subtitle = "Software syntax stabilized into permanent coexistence due to decoupled cognitive and execution constraints.",
    x = "Year (Epoch)",
    y = "Active Repositories Share",
    color = "Syntactic Grammar"
  ) +
  theme_editorial() +
  annotate("label", x = 2018, y = 33, label = "Python/ML Indentation\nResurgence (31%)", 
           size = 2.8, fill = "#ECFDF5", color = "#065F46", fontface = "bold", hjust = 0.8) +
  annotate("label", x = 2018, y = 10, label = "Lisp / Clojure Parens\nEvergreen Niche (7%)", 
           size = 2.8, fill = "#F5F3FF", color = "#5B21B6", fontface = "bold", hjust = 0.8)

title_grob <- textGrob(
  "REPRESENTATIONAL ECONOMICS: EXTINCTION VS. POLYGLOT COEXISTENCE",
  gp = gpar(fontsize = 14, fontface = "bold", col = "#0F172A"),
  x = 0.02, hjust = 0
)
sub_grob <- textGrob(
  "Empirical contrast between zero-sum formal standardization in mathematics and sustainable syntactic diversity in software engineering.",
  gp = gpar(fontsize = 9.5, col = "#475569"),
  x = 0.02, hjust = 0
)
cap_grob <- textGrob(
  "Mathematical Linguistics Group (mlG) • Data: Cambridge Analytical Society Archives (1812), Babbage (1813), GitHub Language Octoverse (1975-2025)",
  gp = gpar(fontsize = 8, col = "#64748B"),
  x = 0.02, hjust = 0
)

fig1 <- arrangeGrob(
  title_grob, sub_grob,
  p_math, p_soft,
  cap_grob,
  ncol = 1,
  heights = c(0.4, 0.3, 5, 5, 0.4)
)

grid.newpage()
grid.draw(fig1)
```

## 3. Hypothesis 5: Lexical Density & The Working Vocabulary Horizon

### Theoretical Context
Linguistic studies of literary corpora traditionally revere William Shakespeare and Herman Melville as historical zeniths of English lexical variety. However, quantitative analysis of hip-hop lyricism demonstrates that the operational constraints of multi-syllabic rhyme schemes, narrative signifying, and cultural neologism demand unprecedented lexical search spaces.
- **Methodology**: Following Matt Daniels (The Pudding), we analyze unique word counts within the first **35,000 words** of an artist's recorded discography or author's prose corpus.
- **Key Benchmarks**:
  - Herman Melville (*Moby Dick*, first 35,000 words): **6,022 unique words**
  - William Shakespeare (Sonnets & Plays, first 35,000 words): **5,170 unique words**
  - Genre Median (Hip-Hop): **~4,200 unique words**

## 4. Hypothesis 4: Semantic Durability (Euphemism Treadmill vs. Vernacular Shields)

### Theoretical Context
- **The Euphemism Treadmill (Pinker, 1994, 2002)**: Bureaucratic and institutional linguistic engineering repeatedly introduces polite, Latinate, or clinical terms to refer to stigmatized conditions. However, because the underlying concept retains social stigma, the stigma quickly migrates from the referent to the newly minted signifier within 15–25 years. This forces subsequent rounds of re-labeling:  
  $$\text{Imbecile (1910)} \longrightarrow \text{Mentally Retarded (1961)} \longrightarrow \text{Special Needs (1988)} \longrightarrow \text{Intellectually Disabled (2007)}$$
- **Organic Vernacular Shields**: In stark contrast, subcultural vernaculars (notably African American Vernacular English and jazz/hip-hop countercultures) employ **enantiosemic inversion** (turning negative semantic primitives like *bad*, *ill*, *sick*, *dope* into signifiers of excellence). This creates a structural linguistic shield: because mainstream outsiders or institutional bureaucracies cannot easily co-opt or formalize the terms without syntactic incongruity, these terms exhibit extraordinary diachronic half-lives (40 to 90+ years) while maintaining high prestige.


```r
# Load Euphemism and Vernacular Durability Data
df_tread <- read.csv("../../data/corpus/euphemism_treadmill.csv")
head(df_tread, 8)
```

```r
# Phase Space Visualization: Half-Life Stability vs. Contemporary Social Valence
p_tread <- ggplot(df_tread, aes(x = half_life_years, y = current_valence, color = category)) +
  geom_hline(yintercept = 0, linetype = "dashed", color = "#94A3B8", linewidth = 0.6) +
  geom_vline(xintercept = 30, linetype = "dotted", color = "#CBD5E1", linewidth = 0.6) +
  geom_point(aes(size = current_valence + 1.2), alpha = 0.85) +
  scale_size_continuous(range = c(3.5, 7.5), guide = "none") +
  scale_color_manual(values = c(
    "Institutional Euphemism" = "#DC2626",
    "Organic Vernacular Shield" = "#059669"
  )) +
  scale_y_continuous(limits = c(-1.05, 1.15), breaks = seq(-1, 1, by = 0.5),
                     labels = c("-1.0 (Pejorated)", "-0.5", "0.0 (Neutral)", "+0.5", "+1.0 (High Prestige)")) +
  scale_x_continuous(limits = c(10, 105), breaks = seq(10, 100, by = 15)) +
  labs(
    title = "THE EUPHEMISM TREADMILL VS. ORGANIC VERNACULAR SHIELDS",
    subtitle = "Diachronic half-life (years of semantic stability) vs. contemporary social valence.
Institutional substitutions decay within 15–25 years, while subcultural enantiosemic inversions maintain multi-decade durability.",
    x = "Linguistic Half-Life (Years of Semantic Stability before Pejoration or Obsolescence)",
    y = "Contemporary Social Valence Score",
    color = "Lexical Engineering Mode",
    caption = "Mathematical Linguistics Group (mlG) • Theoretical Foundations: Pinker (1994, 2002), Smitherman (1977), Labov (1972) • N = 18 diachronic markers"
  ) +
  theme_editorial() +
  annotate("rect", xmin = 10, xmax = 30, ymin = -1.0, ymax = 0.35, alpha = 0.08, fill = "#DC2626") +
  annotate("text", x = 20, y = -0.98, label = "TREADMILL TRAP\nRapid Pejoration & Replacement Cycle", 
           size = 3.2, fontface = "bold", color = "#991B1B", hjust = 0.5) +
  annotate("rect", xmin = 35, xmax = 102, ymin = 0.5, ymax = 1.1, alpha = 0.08, fill = "#059669") +
  annotate("text", x = 68, y = 1.08, label = "VERNACULAR SHIELD\nMulti-Decade Subcultural & Global Prestige Armor", 
           size = 3.2, fontface = "bold", color = "#065F46", hjust = 0.5) +
  geom_label(data = filter(df_tread, term %in% c("Mentally Retarded", "Imbecile / Moron", "Special Needs", "Neurodivergent")),
             aes(label = paste0(term, " (", year_introduced, ")")), 
             nudge_y = -0.09, size = 2.9, fill = "#FEF2F2", color = "#991B1B", fontface = "bold", show.legend = FALSE) +
  geom_label(data = filter(df_tread, term %in% c("Cool", "Bad (as Excellent)", "Dope", "Ill", "Sick", "GOAT")),
             aes(label = paste0(term, " (", year_introduced, ")")), 
             nudge_y = 0.08, size = 2.9, fill = "#ECFDF5", color = "#065F46", fontface = "bold", show.legend = FALSE)

print(p_tread)
```

### Q&A
- **Q: Why does mathematical notation experience total extinction while programming language syntax sustains polyglot coexistence?**  
  **A:** Mathematical notation is bound by a singular global communicational medium (proof publication and unified theorem exchange), rendering incompatible syntaxes zero-sum. Once the Cambridge Analytical Society adopted Leibnizian differentials in 1812, Newtonian fluxions experienced rapid obsolescence (plummeting from a 48% share in 1700 to near 0% for general calculus by 1840). In contrast, software languages decouple execution domains: C-family braces (56% in 2025) dominate system-level deterministic parsing, Python significant whitespace (31%) minimizes cognitive friction in data science and AI orchestration, and Lisp S-expressions (7%) provide homoiconic macro-metaprogramming.
- **Q: Does classical literature represent the upper bound of English lexical diversity?**  
  **A:** No. In a standardized sample of the first 35,000 words of artistic output, underground and golden era hip-hop lyricists systematically surpass classical benchmarks. Aesop Rock records 7,392 unique words (+2,222 unique words, or +43.0% over William Shakespeare's 5,170 words, and +1,370 words over Herman Melville's 6,022 words). Four distinct hip-hop artists (Aesop Rock, GZA, Kool Keith, MF DOOM) exceed Herman Melville's *Moby Dick*, demonstrating that multi-syllabic rhyming structures enforce intense lexical expansion.
- **Q: Why do top-down euphemisms fail to eradicate stigma while subcultural vernacular slang endures for generations?**  
  **A:** Steven Pinker's euphemism treadmill reveals that stigma is conceptual rather than phonetic: institutional substitutes decay within an observed mean half-life of 20.8 years before acquiring pejorative charge (*imbecile* $\rightarrow$ *mentally retarded* $\rightarrow$ *special needs* $\rightarrow$ *intellectually disabled*). Conversely, subcultural enantiosemic inversions (*cool*, *bad*, *ill*, *dope*) demonstrate an observed mean half-life of 50.8 years (reaching 93 years for *cool*) because syntactic and cultural barriers insulate the inverted prestige signifier from bureaucratic dilution.

### Data Analysis Key Findings
- **Shannon Entropy of Syntactic Diversity**: Mathematics collapsed from a high-entropy competitive state in 1700 ($H = 0.692$) to a near-zero entropy monopoly by 1900 ($H = 0.098$, with Leibnizian notation commanding 98.0% of calculus literature). In software syntax, entropy has remained sustainably high across 50 years ($H = 1.054$ in 2025), preserving a four-paradigm ecosystem.
- **Underground Lyricism Threshold**: 10 out of 25 sampled hip-hop discographies exceed William Shakespeare's canonical working vocabulary of 5,170 words per 35,000-word corpus. The underground abstract tradition averages 6,599 unique words, outperforming both the genre median (4,200 words) and canonical English literature.
- **Half-Life Disparity in Lexical Engineering**: Institutional euphemisms exhibit a maximum observed half-life of only 28.0 years (*underprivileged* and *neurodivergent*), with a mean contemporary social valence of $-0.291$. In contrast, organic vernacular shields exhibit a minimum half-life of 21.0 years (*fire*) and an average of 50.8 years, retaining a strongly positive mean valence of $+0.861$.
